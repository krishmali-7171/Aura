```python

# ============================================================
# PROJECT: ReAct-Based Intelligent Placement Preparation Agent
# SUBJECT: Agentic AI
# UNIT 2: Reasoning, Planning and Decision-Making (CO-2)
# FRAMEWORK: ReAct (Reason + Act)
# ============================================================

# STEP 1: DEFINE THE MAIN GOAL

goal = "Prepare a 3-day placement preparation schedule"

print("=" * 65)
print("       ReAct-BASED INTELLIGENT PLACEMENT PREPARATION AGENT")
print("=" * 65)

print("\nMAIN GOAL:")
print(goal)


# STEP 2: TASK DECOMPOSITION

tasks = [
    {"name": "SQL Preparation", "hours": 2},
    {"name": "Python OOP Preparation", "hours": 2},
    {"name": "System Design Preparation", "hours": 2}
]

print("\n[THOUGHT] I need to divide the main goal into smaller tasks.")

print("\n[ACTION] Breaking the goal into subtasks.")

print("\n[OBSERVATION] The following subtasks were identified:")

for i, task in enumerate(tasks, 1):
    print(f"{i}. {task['name']} - {task['hours']} hours")


# STEP 3: DEFINE AVAILABLE STUDY TIME

availability = {
    1: 2,
    2: 1,
    3: 3
}

print("\n[THOUGHT] I should check the available study time.")

print("[ACTION] Checking daily time availability.")

for day, hours in availability.items():
    print(f"Day {day}: {hours} hours available")


# STEP 4: CREATE THE INITIAL PLAN

plan = {
    1: [
        {"name": "SQL Preparation", "hours": 2}
    ],

    2: [
        {"name": "Python OOP Preparation", "hours": 2}
    ],

    3: [
        {"name": "System Design Preparation", "hours": 2}
    ]
}

print("\n" + "=" * 65)
print("                  INITIAL PLAN")
print("=" * 65)

for day, day_tasks in plan.items():
    print(f"\nDay {day}:")

    for task in day_tasks:
        print(f"  - {task['name']} ({task['hours']} hours)")


# STEP 5: CREATE TOOLS FOR THE AGENT

# Tool 1: Check available time

def check_available_time(day):

    return availability[day]


# Tool 2: Execute a task

def execute_task(task, available_time):

    required_time = task["hours"]

    if available_time >= required_time:

        return True, "Task completed successfully"

    else:

        return False, (
            f"Task failed! Required {required_time} hours, "
            f"but only {available_time} hours are available."
        )


# Tool 3: Dynamic replanning

def replan_schedule():

    print("\n" + "=" * 65)
    print("                 DYNAMIC REPLANNING")
    print("=" * 65)

    print("\n[THOUGHT] The original Python OOP task requires 2 hours.")
    print("[THOUGHT] Only 1 hour is available on Day 2.")
    print("[THOUGHT] I should divide the task into smaller subtasks.")

    print("\n[ACTION] Modifying the original schedule.")

    revised_plan = {
        1: [
            {"name": "SQL Preparation", "hours": 2}
        ],

        2: [
            {"name": "Python OOP - Fundamentals", "hours": 1}
        ],

        3: [
            {"name": "Python OOP - Practice", "hours": 1},
            {"name": "System Design Preparation", "hours": 2}
        ]
    }

    print("\n[OBSERVATION] The revised schedule has been created.")

    return revised_plan


# STEP 6: REACT EXECUTION ENGINE

print("\n" + "=" * 65)
print("                 REACT EXECUTION TRACE")
print("=" * 65)

replanning_done = False

for day in range(1, 4):

    print("\n" + "-" * 65)
    print(f"                       DAY {day}")
    print("-" * 65)

    print("\n[THOUGHT] I need to execute the tasks scheduled for today.")

    print("[ACTION] Checking available study time.")

    available_time = check_available_time(day)

    print("[OBSERVATION]", available_time, "hours available.")

    # Execute tasks in the current plan

    current_tasks = plan[day]

    for task in current_tasks:

        print("\n[THOUGHT] I should complete:", task["name"])

        print("[ACTION] Attempting to execute:", task["name"])

        success, result = execute_task(task, available_time)

        print("[OBSERVATION]", result)

        if success:

            available_time -= task["hours"]

            print("[THOUGHT] Task completed. Moving forward.")

        else:

            # Trigger dynamic replanning only once

            if day == 2 and not replanning_done:

                print("\n[THOUGHT] The original plan is not feasible.")

                print("[ACTION] Triggering dynamic replanning.")

                plan = replan_schedule()

                replanning_done = True

                # Execute the revised task for the current day

                print("\n[ACTION] Executing the revised Day 2 task.")

                revised_task = plan[day][0]

                success, result = execute_task(
                    revised_task,
                    availability[day]
                )

                print("[OBSERVATION]", result)

                if success:
                    print("[THOUGHT] Revised Day 2 task completed.")

            else:

                print("[THOUGHT] Task could not be completed.")


# STEP 7: VALIDATE THE FINAL PLAN

print("\n" + "=" * 65)
print("                 FINAL PLAN VALIDATION")
print("=" * 65)

valid = True

for day, day_tasks in plan.items():

    total_hours = sum(task["hours"] for task in day_tasks)

    available_hours = availability[day]

    print(f"\nDay {day}:")
    print("Scheduled hours:", total_hours)
    print("Available hours:", available_hours)

    if total_hours <= available_hours:

        print("[OBSERVATION] Schedule is valid.")

    else:

        print("[OBSERVATION] Scheduling conflict detected.")

        valid = False


# STEP 8: DISPLAY THE FINAL REVISED SCHEDULE

print("\n" + "=" * 65)
print("                 FINAL REVISED SCHEDULE")
print("=" * 65)

for day, day_tasks in plan.items():

    print(f"\nDay {day}:")

    for task in day_tasks:

        print(f"  - {task['name']} ({task['hours']} hours)")


# STEP 9: FINAL RESULT

print("\n" + "=" * 65)
print("                    FINAL RESULT")
print("=" * 65)

if valid and replanning_done:

    print("\nAgent successfully completed the planning process.")

    print("Dynamic replanning was successfully demonstrated.")

    print("The revised schedule fits the available study time.")

elif valid:

    print("\nAgent completed the planning process.")

else:

    print("\nAgent could not create a fully valid schedule.")

print("\nReAct Cycle: Thought -> Action -> Observation")

print("\nPROJECT EXECUTION COMPLETED.")

print("=" * 65)
```
