```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
from torchvision.utils import make_grid
from IPython.display import Audio, display

# Reproducibility
torch.manual_seed(42)
np.random.seed(42)
random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5),
                         (0.5, 0.5, 0.5))
])

dataset = datasets.CIFAR10(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

# A smaller subset makes experimentation easier.
# The DCGAN below still trains for the required 50 epochs.
subset = Subset(dataset, range(10000))
batch_size = 128

train_loader = DataLoader(
    subset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

print("Training images:", len(subset))
print("Batch size:", batch_size)
```

```python
LATENT_DIM = 100

class Generator(nn.Module):
    def __init__(self, latent_dim=100):
        super().__init__()

        self.net = nn.Sequential(
            nn.ConvTranspose2d(latent_dim, 256, 4, 1, 0, bias=False),
            nn.BatchNorm2d(256),
            nn.ReLU(True),

            nn.ConvTranspose2d(256, 128, 4, 2, 1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(True),

            nn.ConvTranspose2d(128, 64, 4, 2, 1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(True),

            nn.ConvTranspose2d(64, 3, 4, 2, 1, bias=False),
            nn.Tanh()
        )

    def forward(self, z):
        return self.net(z)


class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv2d(3, 64, 4, 2, 1),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(64, 128, 4, 2, 1, bias=False),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(128, 256, 4, 2, 1, bias=False),
            nn.BatchNorm2d(256),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(256, 1, 4, 1, 0),
            nn.Sigmoid()
        )

    def forward(self, image):
        return self.net(image).view(-1)


def create_dcgan(latent_dim=100, lr=0.0002):
    G = Generator(latent_dim).to(device)
    D = Discriminator().to(device)

    optimizer_G = optim.Adam(
        G.parameters(), lr=lr, betas=(0.5, 0.999)
    )
    optimizer_D = optim.Adam(
        D.parameters(), lr=lr, betas=(0.5, 0.999)
    )

    return G, D, optimizer_G, optimizer_D
```

```python
def show_generated_images(G, latent_dim, title):
    G.eval()

    with torch.no_grad():
        z = torch.randn(25, latent_dim, 1, 1, device=device)
        images = G(z).cpu()
        images = (images + 1) / 2

    grid = make_grid(images, nrow=5)
    plt.figure(figsize=(7, 7))
    plt.imshow(grid.permute(1, 2, 0).clamp(0, 1))
    plt.title(title)
    plt.axis("off")
    plt.show()

    G.train()


G, D, opt_G, opt_D = create_dcgan(
    latent_dim=LATENT_DIM,
    lr=0.0002
)

criterion = nn.BCELoss()
epochs = 50

loss_G_history = []
loss_D_history = []
saved_grids = {}

for epoch in range(1, epochs + 1):
    total_G = 0
    total_D = 0

    for real_images, _ in train_loader:
        real_images = real_images.to(device)
        current_batch = real_images.size(0)

        real_labels = torch.ones(current_batch, device=device)
        fake_labels = torch.zeros(current_batch, device=device)

        # Train discriminator
        opt_D.zero_grad()

        real_output = D(real_images)
        loss_real = criterion(real_output, real_labels)

        noise = torch.randn(
            current_batch, LATENT_DIM, 1, 1, device=device
        )
        fake_images = G(noise)

        fake_output = D(fake_images.detach())
        loss_fake = criterion(fake_output, fake_labels)

        loss_D = (loss_real + loss_fake) / 2
        loss_D.backward()
        opt_D.step()

        # Train generator
        opt_G.zero_grad()

        generated_output = D(fake_images)
        loss_G = criterion(generated_output, real_labels)

        loss_G.backward()
        opt_G.step()

        total_G += loss_G.item()
        total_D += loss_D.item()

    avg_G = total_G / len(train_loader)
    avg_D = total_D / len(train_loader)

    loss_G_history.append(avg_G)
    loss_D_history.append(avg_D)

    print(
        f"Epoch {epoch:02d}/50 | "
        f"Generator Loss: {avg_G:.4f} | "
        f"Discriminator Loss: {avg_D:.4f}"
    )

    if epoch in [10, 30, 50]:
        show_generated_images(
            G, LATENT_DIM, f"DCGAN Generated Images - Epoch {epoch}"
        )

        G.eval()
        with torch.no_grad():
            z = torch.randn(25, LATENT_DIM, 1, 1, device=device)
            saved_grids[epoch] = (
                (G(z).cpu() + 1) / 2
            )
        G.train()

print("DCGAN training completed!")
```

```python
class VAE(nn.Module):
    def __init__(self, latent_dim=128):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(3 * 32 * 32, 512),
            nn.ReLU()
        )

        self.mu = nn.Linear(512, latent_dim)
        self.logvar = nn.Linear(512, latent_dim)

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, 3 * 32 * 32),
            nn.Tanh()
        )

    def forward(self, x):
        h = self.encoder(x)
        mu = self.mu(h)
        logvar = self.logvar(h)

        std = torch.exp(0.5 * logvar)
        z = mu + std * torch.randn_like(std)

        reconstruction = self.decoder(z).view(-1, 3, 32, 32)
        return reconstruction, mu, logvar


def vae_loss(reconstruction, original, mu, logvar):
    reconstruction_loss = nn.functional.mse_loss(
        reconstruction, original, reduction="sum"
    )

    kl_loss = -0.5 * torch.sum(
        1 + logvar - mu.pow(2) - logvar.exp()
    )

    return reconstruction_loss + kl_loss


vae = VAE(latent_dim=128).to(device)
vae_optimizer = optim.Adam(vae.parameters(), lr=0.001)

vae_epochs = 5  # Increase to 10 or more if time permits.

for epoch in range(1, vae_epochs + 1):
    vae.train()
    total_loss = 0

    for images, _ in train_loader:
        images = images.to(device)
        vae_optimizer.zero_grad()

        reconstruction, mu, logvar = vae(images)
        loss = vae_loss(reconstruction, images, mu, logvar)

        loss.backward()
        vae_optimizer.step()

        total_loss += loss.item()

    print(
        f"VAE Epoch {epoch}/{vae_epochs}, "
        f"Loss per image: {total_loss / len(subset):.4f}"
    )

# Generate images from random latent vectors
vae.eval()

with torch.no_grad():
    z = torch.randn(25, 128, device=device)
    generated = vae.decoder(z).view(-1, 3, 32, 32).cpu()
    generated = (generated + 1) / 2

grid = make_grid(generated, nrow=5)
plt.figure(figsize=(7, 7))
plt.imshow(grid.permute(1, 2, 0).clamp(0, 1))
plt.title("Images Generated by VAE")
plt.axis("off")
plt.show()
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
text_model = AutoModelForCausalLM.from_pretrained(model_name)
text_model.to(device)
text_model.eval()

def generate_text(prompt, temperature=0.8, top_k=50):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        output = text_model.generate(
            **inputs,
            max_new_tokens=60,
            do_sample=True,
            temperature=temperature,
            top_k=top_k,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(output[0], skip_special_tokens=True)


prompt = "A creative product description for a smart learning assistant:"

settings = [
    (0.5, 10),
    (0.5, 50),
    (0.8, 10),
    (0.8, 50),
    (1.2, 50)
]

text_samples = []

for i, (temperature, top_k) in enumerate(settings, start=1):
    result = generate_text(prompt, temperature, top_k)
    text_samples.append(result)

    print(f"\nSample {i}: Temperature={temperature}, Top-k={top_k}")
    print(result)
    print("-" * 60)
```

```python
import pretty_midi
import wave
import struct
import math

sample_rate = 22050
bpm = 100
beat = 60 / bpm

# Four bars, four beats per bar
notes = [60, 64, 67, 72, 67, 64, 62, 65,
         69, 74, 72, 67, 64, 62, 60, 60]

midi = pretty_midi.PrettyMIDI(initial_tempo=bpm)
instrument = pretty_midi.Instrument(program=0)

for i, pitch in enumerate(notes):
    start = i * beat
    end = start + beat * 0.9

    note = pretty_midi.Note(
        velocity=90,
        pitch=pitch,
        start=start,
        end=end
    )
    instrument.notes.append(note)

midi.instruments.append(instrument)
midi.write("generated_melody.mid")

# Synthesize a basic piano-like tone directly into a WAV file.
# This does not require an external soundfont.
duration = len(notes) * beat
audio = np.zeros(int(sample_rate * duration))

for i, pitch in enumerate(notes):
    start = i * beat
    end = start + beat * 0.9

    first = int(start * sample_rate)
    last = min(int(end * sample_rate), len(audio))
    t = np.arange(last - first) / sample_rate

    frequency = 440 * (2 ** ((pitch - 69) / 12))
    tone = (
        np.sin(2 * np.pi * frequency * t)
        + 0.3 * np.sin(4 * np.pi * frequency * t)
    )

    envelope = np.minimum(1, t * 20) * np.maximum(
        0, 1 - t / max(end - start, 0.001)
    )

    audio[first:last] += 0.25 * tone * envelope

audio = audio / max(np.max(np.abs(audio)), 1e-8)
audio_int16 = (audio * 32767).astype(np.int16)

with wave.open("generated_melody.wav", "wb") as wav_file:
    wav_file.setnchannels(1)
    wav_file.setsampwidth(2)
    wav_file.setframerate(sample_rate)
    wav_file.writeframes(audio_int16.tobytes())

print("MIDI saved: generated_melody.mid")
print("Audio saved: generated_melody.wav")

display(Audio(audio, rate=sample_rate))
```

```python
# Small diagnostic tuning runs
# Each configuration trains for only one epoch on 1,000 images.

import pandas as pd

small_subset = Subset(dataset, range(1000))

results = []

experiments = [
    ("Learning rate", 0.0001, 100, 64),
    ("Learning rate", 0.0002, 100, 64),
    ("Learning rate", 0.0004, 100, 64),

    ("Latent dimension", 100, 100, 64),
    ("Latent dimension", 64, 64, 64),
    ("Latent dimension", 128, 128, 64),

    ("Batch size", 64, 100, 64),
    ("Batch size", 128, 100, 128),
    ("Batch size", 256, 100, 256)
]

# The second field is the learning rate for the learning-rate experiments.
# For latent-dimension and batch-size experiments, use lr=0.0002.
for parameter, value, latent_dim, batch in experiments:
    lr = value if parameter == "Learning rate" else 0.0002

    # For batch-size experiments, use the specified batch size.
    loader = DataLoader(
        small_subset,
        batch_size=batch,
        shuffle=True,
        num_workers=2,
        pin_memory=torch.cuda.is_available()
    )

    gen, disc, og, od = create_dcgan(latent_dim, lr)
    criterion = nn.BCELoss()

    gen_loss_total = 0
    disc_loss_total = 0

    for real, _ in loader:
        real = real.to(device)
        n = real.size(0)

        ones = torch.ones(n, device=device)
        zeros = torch.zeros(n, device=device)

        # Discriminator
        od.zero_grad()
        real_loss = criterion(disc(real), ones)

        z = torch.randn(n, latent_dim, 1, 1, device=device)
        fake = gen(z)
        fake_loss = criterion(disc(fake.detach()), zeros)

        d_loss = (real_loss + fake_loss) / 2
        d_loss.backward()
        od.step()

        # Generator
        og.zero_grad()
        g_loss = criterion(disc(fake), ones)
        g_loss.backward()
        og.step()

        disc_loss_total += d_loss.item()
        gen_loss_total += g_loss.item()

    results.append({
        "Parameter": parameter,
        "Value": value if parameter != "Batch size" else batch,
        "Generator Loss": round(gen_loss_total / len(loader), 4),
        "Discriminator Loss": round(disc_loss_total / len(loader), 4)
    })

    print(f"Completed: {parameter} = {value}")

tuning_table = pd.DataFrame(results)
display(tuning_table)
```

```python
import gradio as gr
from torchvision.transforms.functional import to_pil_image

def image_module():
    G.eval()
    with torch.no_grad():
        z = torch.randn(1, LATENT_DIM, 1, 1, device=device)
        img = G(z).cpu().squeeze(0)
        img = (img + 1) / 2
    return to_pil_image(img.clamp(0, 1))


def text_module(prompt, temperature, top_k):
    return generate_text(prompt, temperature, int(top_k))


def music_module():
    return "generated_melody.wav"


with gr.Blocks(title="Multi-Modal Creative Studio") as demo:
    gr.Markdown("# Multi-Modal Generative AI Studio")
    gr.Markdown("Generate images, creative text, and music.")

    with gr.Tab("Image Generation"):
        gr.Markdown("Generate a new image using the trained DCGAN.")
        image_button = gr.Button("Generate Image")
        image_output = gr.Image(label="Generated Image")
        image_button.click(
            fn=image_module,
            outputs=image_output
        )

    with gr.Tab("Text Generation"):
        prompt_input = gr.Textbox(
            label="Enter a prompt",
            value="A creative product description for a smart learning assistant:"
        )
        temperature_input = gr.Slider(
            0.5, 1.2, value=0.8, step=0.1,
            label="Temperature"
        )
        top_k_input = gr.Dropdown(
            choices=[10, 50],
            value=50,
            label="Top-k"
        )
        text_button = gr.Button("Generate Text")
        text_output = gr.Textbox(
            label="Generated Text",
            lines=6
        )
        text_button.click(
            fn=text_module,
            inputs=[
                prompt_input,
                temperature_input,
                top_k_input
            ],
            outputs=text_output
        )

    with gr.Tab("Music Generation"):
        gr.Markdown("Play the generated four-bar melody.")
        music_button = gr.Button("Play Generated Music")
        music_output = gr.Audio(
            label="Generated Melody",
            type="filepath"
        )
        music_button.click(
            fn=music_module,
            outputs=music_output
        )

demo.launch(share=True)
```
